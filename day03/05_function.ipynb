```python
def get_sum(a, b):
    return a + b

a, b = 10, 20

print(f"두수합 => {get_sum(a, b)}")
```

```python
role = """
세개의 점수를 매개변수로 전달받아 세수의 합을 구해서 반환하는 함수를 만들고
호출해 보세요
"""
print(role)
def get_sum(a, b, c):
    return a + b + c
a, b, c = 10, 20, 30
print(f"세수합 => {get_sum(a, b, c)}")
```

```python
role = """
가로, 세로길이를 매개변수로 전달받아 사각형의 넓이를 구해서 출력하는 함수를
만들고(return 문 없음) 호출해서 사용해 보세요
"""
print(role)

def area_calculation(w, h):
    print(f"가로길이: {w}, 세로길이: {h}, 사각형의 넓이: {w * h}")

area_calculation(10, 20)
```

```python
role = """
정수 하나를 매개변수로 전달받아 짝수인지 홀수인지 문자열로 리턴하는 함수를
만들고 사용해 보세요.
예.) a = is_check(10)
     print(a) => 출력결과 짝수
"""
print(role)
def check_even_odd(a):
    return "짝수" if a % 2 == 0 else "홀수"
a = int(input("숫자를 입력하세요: "))
print(f"숫자 {a} 은/는 {check_even_odd(a)} 입니다.")
```

```python
print("===> 람다")
add_lambda = lambda a, b: a + b
print(add_lambda(10, 20))

check_even_odd_lambda = lambda a: "짝수" if a % 2 == 0 else "홀수"
print(check_even_odd_lambda(7))
```

```python
print("===> 예시 2 - sorted()와 람다 함께 사용")
students = [("철수", 85), ("영희", 92), ("민수", 78)]

sorted_students = sorted(students, key=lambda x: x[1], reverse=1)
print(sorted_students)
```

```python
print("===> 여러 값을 한 번에 반환 (튜플 반환)")
def min_max(numbers):
    return min(numbers), max(numbers)

numbers = [4, 1, 9, 2, 7]
low, high = min_max(numbers)
print(f"{numbers} low, high: {low}, {high}")
```

```python
print("===> remove 테스트")
test_numbers = [4, 1, 3, 4, 5]
test_numbers.remove(4)
print(test_numbers)
print("첫번째 타겟만 삭제되었음.")

print("전체삭제: return [value for value in values if value != target]")
def remove_all(values, target):
    return [value for value in values if value != target]

print(remove_all(test_numbers, 4))
```

```python
print("===> 가변 인자 *args")
def total_sum(*args):
    return sum(args)

print(total_sum(1, 2, 3))
print(total_sum(1, 2, 3, 4, 5))
```
